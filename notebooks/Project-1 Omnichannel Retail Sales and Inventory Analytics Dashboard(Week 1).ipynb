```python
#Import Libraries
import pandas as pd
import numpy as np
```

```python
#Load dataset
df = pd.read_csv(r'C:\Users\ADMIN\Downloads\archive\retail_sales_dataset.csv')

df.head()
```

```python
#Basic info
df.info()
df.describe()
```

```python
#Check Missing Values
df.isnull().sum()
```

```python
#Handle Missing Values
df['Price per Unit'] = df['Price per Unit'].fillna(df['Price per Unit'].mean())
df['Quantity'] = df['Quantity'].fillna(0)
```

```python
#Remove Duplicates
df.duplicated().sum()

df = df.drop_duplicates()
```

```python
#Clean Column Names
df.columns = df.columns.str.lower().str.strip()

df.columns
```

```python
#Fix Data Types
df['date'] = pd.to_datetime(df['date'])

df['quantity'] = df['quantity'].astype(int)
df['price per unit'] = df['price per unit'].astype(float)
```

```python
#Create New Columns
df['total_sales'] = df['quantity'] * df['price per unit']

df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['hour'] = df['date'].dt.hour
df['weekday'] = df['date'].dt.day_name()
```

```python
#Validate Data
df.head()
df.describe()
```

```python
# Total revenue
print("Total Sales:", df['total_sales'].sum())

# Top products
df.groupby('product category')['total_sales'].sum().sort_values(ascending=False).head()
```

```python
#Save Cleaned Data
df.to_csv('cleaned_retail_data.csv', index=False)
```
